```python
# Import Required Libraries
import os
from pathlib import Path
from typing import List

# List Files in materiais Folder
materials_dir = Path('materiais')
files = sorted(materials_dir.glob('*'))
files

# Read Material Information
entries = []
for path in files:
    if path.is_file():
        entries.append({
            'name': path.name,
            'type': path.suffix.lower(),
            'size': path.stat().st_size,
        })
entries

# Build Course Content Entries
content_lines = ["# Conteúdo do Curso", "", "## Materiais", ""]
for entry in entries:
    content_lines.append(f"- {entry['name']} ({entry['type']})")
content_lines

# Write readme.md with Course Content
readme_path = Path('README.md')
readme_path.write_text('\n'.join(content_lines), encoding='utf-8')
readme_path.read_text(encoding='utf-8')
```
